````python
# ============================================================
# 第 1 格：安裝套件、讀取金鑰、設定模型
# ============================================================
!pip install -q groq gradio

import sqlite3
import pandas as pd
from IPython.display import display
from google.colab import userdata
from groq import Groq
import gradio as gr

# 🔴 金鑰只能從左側「🔑 密鑰（Secrets）」讀取，絕對不可以寫成 api_key="gsk_..."
#    寫死的金鑰一旦存到 GitHub，就等於公開給全世界使用。
#    也不要 print 出金鑰，因為輸出結果也會一起存到 GitHub。
try:
    client = Groq(api_key=userdata.get('GROQ_API_KEY'))
except Exception as e:
    print("❌ 讀不到金鑰。請檢查：")
    print("   1. 左側 🔑 是否新增了名稱為 GROQ_API_KEY 的密鑰（大小寫要完全一樣）")
    print("   2. 該密鑰的「筆記本存取權」開關是否已打開")
    raise

# 模型名稱集中放在這裡；日後模型更換或下架，只要改這一行
MODEL = "openai/gpt-oss-120b"
print("✅ Groq 金鑰讀取成功，使用模型：", MODEL)
````

````python
# ============================================================
# 第 2 格：建立資料庫與初始資料
# 提醒：每次執行都要 DROP TABLE IF EXISTS 再重建
# ============================================================
# （把 AI 產生、你看過並理解的程式貼在這裡）
# (2) 建立資料庫與初始資料
import sqlite3

def init_db():
    """每次執行時先刪除舊資料表，重建兩張資料表並寫入初始資料"""
    conn = sqlite3.connect('clinic.db')
    cursor = conn.cursor()
    try:
        # 刪除舊表確保乾淨
        cursor.execute("DROP TABLE IF EXISTS registrations")
        cursor.execute("DROP TABLE IF EXISTS departments")

        # 1. 建立科別資料表（包含名稱、每日最大名額、當日已掛號人數）
        cursor.execute("""
            CREATE TABLE departments (
                id INTEGER PRIMARY KEY AUTOINCREMENT,
                name TEXT NOT NULL,
                max_slots INTEGER NOT NULL,
                current_booked INTEGER NOT NULL
            )
        """)

        # 2. 建立掛號紀錄資料表
        cursor.execute("""
            CREATE TABLE registrations (
                id INTEGER PRIMARY KEY AUTOINCREMENT,
                patient_name TEXT NOT NULL,
                dept_name TEXT NOT NULL,
                status TEXT NOT NULL
            )
        """)

        # 3. 放入初始資料（內科、外科、牙科，預設每日限額 5 人）
        initial_depts = [
            ('內科', 5, 0),
            ('外科', 5, 0),
            ('牙科', 5, 0)
        ]
        cursor.executemany("INSERT INTO departments (name, max_slots, current_booked) VALUES (?, ?, ?)", initial_depts)

        conn.commit()
    finally:
        conn.close()

# 執行初始化
init_db()
print("資料庫建立成功，初始科別資料已載入！")
````

````python
# ============================================================
# 第 3 格：顯示資料表
# 提醒：定義 show_tables() 並呼叫一次
# ============================================================
# （把 AI 產生、你看過並理解的程式貼在這裡）
# (3) 顯示資料表
import pandas as pd

def show_tables():
    """使用 pandas 讀取並顯示資料庫中的兩張表"""
    conn = sqlite3.connect('clinic.db')
    try:
        print("=== 科別資料表 (departments) ===")
        df_dept = pd.read_sql_query("SELECT * FROM departments", conn)
        display(df_dept)

        print("\n=== 掛號紀錄表 (registrations) ===")
        df_reg = pd.read_sql_query("SELECT * FROM registrations", conn)
        display(df_reg)
    finally:
        conn.close()

show_tables()
````

````python
# ============================================================
# 第 4 格：AI 解讀函式
# 提醒：回傳 (動作, 科別, 原始回覆)；不可直接相信 AI 的回覆
# ============================================================
# （把 AI 產生、你看過並理解的程式貼在這裡）
# (4) AI 解讀函式
def interpret_intent(user_input):
    """
    呼叫 Groq API 解析使用者的話，回傳 (動作, 對象)
    動作限制：掛號, 取消, 查詢, 未知
    對象限制：內科, 外科, 牙科, 未知
    """
    prompt = f"""
你是一個診所語意解析助理。請分析使用者的話，並嚴格依照格式回傳「動作,對象」。

規則：
1. 動作只能是以下其中之一：掛號, 取消, 查詢, 未知
2. 對象只能是以下其中之一：內科, 外科, 牙科, 未知
3. 絕對不可以提供任何醫療建議、病情分析或推薦科別。如果使用者僅描述症狀（例如發燒、咳嗽、骨折、肚子痛）而沒有明確指定內科、外科或牙科，一律將對象輸出為「未知」。
4. 輸出格式必須為純文字「動作,對象」，例如：`掛號,內科` 或 `未知,未知`。不要包含其他多餘的字元。

使用者說的話是：{user_input}
"""

    try:
        response = client.chat.completions.create(
            model=MODEL,
            messages=[
                {"role": "system", "content": "你是一個精準的語意分類器，只輸出指定的 2 個詞彙。"},
                {"role": "user", "content": prompt}
            ],
            temperature=0  # 設為 0 以確保輸出穩定
        )
        raw_text = response.choices[0].message.content.strip()
    except Exception as e:
        # API 失敗時回傳預設未知，不可讓程式當掉
        raw_text = "未知,未知"

    # 安全防護：從 AI 回覆中嚴格過濾與清洗允許清單的值
    valid_actions = ["掛號", "取消", "查詢", "未知"]
    valid_targets = ["內科", "外科", "牙科", "未知"]

    action, target = "未知", "未知"
    if "," in raw_text:
        parts = raw_text.split(",")
        if len(parts) >= 2:
            potential_action = parts[0].strip()
            potential_target = parts[1].strip()

            if potential_action in valid_actions:
                action = potential_action
            if potential_target in valid_targets:
                target = potential_target

    return action, target, raw_text
````

````python
# ============================================================
# 第 5 格：測試清單
# 提醒：至少 8 句，要與規格書第 7 節一致
# ============================================================
# （把 AI 產生、你看過並理解的程式貼在這裡）
# (5) 測試清單執行與驗證
def run_test_suite():
    """執行至少 8 句測試案例（包含正常、錯誤、邊界），並印出結果與正確率統計"""
    test_cases = [
        # (類型, 使用者說的話, 預期動作, 預期對象)
        ("正常", "幫我掛號內科", "掛號", "內科"),
        ("正常", "請問外科還有剩餘名額嗎？", "查詢", "外科"),
        ("正常", "我要取消牙科掛號", "取消", "牙科"),
        ("錯誤", "我最近一直咳嗽發燒", "未知", "未知"),
        ("邊界", "請幫我掛號", "未知", "未知"),
        ("正常", "我想查詢內科剩餘名額", "查詢", "內科")
    ]

    passed_count = 0
    print("========== 開始執行 AI 語意測試清單 ==========\n")

    for i, (test_type, sentence, expected_action, expected_target) in enumerate(test_cases, 1):
        action, target, raw_text = interpret_intent(sentence)

        is_passed = (action == expected_action and target == expected_target)
        if is_passed:
            passed_count += 1
            status_str = "【通過 ✅】"
        else:
            status_str = "【失敗 ❌】"

        print(f"測試 #{i} [{test_type}] {status_str}")
        print(f"  - 使用者語句: {sentence}")
        print(f"  - AI 原始回覆: {raw_text}")
        print(f"  - 系統解讀結果: 動作={action}, 對象={target}")
        print(f"  - 預期結果: 動作={expected_action}, 對象={expected_target}\n")

    print(f"測試總結：共 {len(test_cases)} 句，成功通過 {passed_count} 句。")

run_test_suite()
````

````python
# ============================================================
# 第 6 格：功能函式（掛號、取消、查詢）
# 提醒：fetchone() 後先檢查 None；用 try/finally 關閉連線
# ============================================================
# （把 AI 產生、你看過並理解的程式貼在這裡）
# (6) 功能函式
def process_clinic_request(user_input, patient_name="王小明"):
    """
    核心業務邏輯處理函式：
    結合 AI 意圖解析，並根據動作與科別操作 SQLite 資料庫
    """
    action, target, raw_text = interpret_intent(user_input)

    # 1. 檢查是否無法辨識或僅描述症狀
    if action == "未知" or target == "未知":
        return f"【AI 原始回覆】{raw_text}\n【系統回應】無法辨識您的需求或科別。本系統僅提供內科、外科、牙科服務，且不提供任何醫療建議，請明確指定欲掛號、取消或查詢的科別。"

    conn = sqlite3.connect('clinic.db')
    cursor = conn.cursor()
    try:
        # 檢查該科別是否存在資料庫中
        cursor.execute("SELECT max_slots, current_booked FROM departments WHERE name = ?", (target,))
        row = cursor.fetchone() # fetchone() 之後一定要檢查是否為 None

        if row is None:
            return f"【AI 原始回覆】{raw_text}\n【系統回應】查無此科別（{target}），本診所僅提供：內科、外科、牙科。"

        max_slots, current_booked = row

        # 2. 處理「查詢」功能
        if action == "查詢":
            remaining = max_slots - current_booked
            return f"【AI 原始回覆】{raw_text}\n【系統回應】[{target}] 當日總名額：{max_slots} 人，已掛號：{current_booked} 人，剩餘名額：{remaining} 人。"

        # 3. 處理「掛號」功能
        elif action == "掛號":
            # 檢查名額是否已滿（邊界情況：剩餘 0）
            if current_booked >= max_slots:
                return f"【AI 原始回覆】{raw_text}\n【系統回應】掛號失敗！[{target}] 今日名額已滿（上限 {max_slots} 人）。"

            # 檢查該病患是否已經在該科別重複掛號且狀態有效
            cursor.execute("SELECT id FROM registrations WHERE patient_name = ? AND dept_name = ? AND status = '有效'", (patient_name, target))
            existing_reg = cursor.fetchone()
            if existing_reg is not None:
                return f"【AI 原始回覆】{raw_text}\n【系統回應】掛號失敗！您（{patient_name}）已經掛過 [{target}] 了，請勿重複掛號。"

            # 執行掛號：更新已預約人數並新增紀錄
            cursor.execute("UPDATE departments SET current_booked = current_booked + 1 WHERE name = ?", (target,))
            cursor.execute("INSERT INTO registrations (patient_name, dept_name, status) VALUES (?, ?, '有效')", (patient_name, target))
            conn.commit()

            return f"【AI 原始回覆】{raw_text}\n【系統回應】掛號成功！已為您完成 [{target}] 掛號（患者：{patient_name}）。"

        # 4. 處理「取消」功能
        elif action == "取消":
            # 檢查是否有該病患的有效掛號紀錄
            cursor.execute("SELECT id FROM registrations WHERE patient_name = ? AND dept_name = ? AND status = '有效'", (patient_name, target))
            reg_row = cursor.fetchone()

            if reg_row is None:
                return f"【AI 原始回覆】{raw_text}\n【系統回應】取消失敗！查無您（{patient_name}）在 [{target}] 的有效掛號紀錄。"

            reg_id = reg_row[0]
            # 更新紀錄狀態為已取消，並將科別已掛號人數 -1
            cursor.execute("UPDATE registrations SET status = '已取消' WHERE id = ?", (reg_id,))
            cursor.execute("UPDATE departments SET current_booked = current_booked - 1 WHERE name = ?", (target,))
            conn.commit()

            return f"【AI 原始回覆】{raw_text}\n【系統回應】取消成功！已為您取消 [{target}] 的掛號紀錄。"

        else:
            return f"【AI 原始回覆】{raw_text}\n【系統回應】不支援的動作指令。"

    finally:
        conn.close() # 透過 try/finally 確保資料庫連線一定會被關閉
````

````python
# ============================================================
# 第 7 格：主流程與完整流程測試
# 提醒：最後呼叫 show_tables() 確認資料庫有變化
# ============================================================
# （把 AI 產生、你看過並理解的程式貼在這裡）
# (7) 主流程與完整流程測試
def run_integration_test():
    """完整模擬使用者互動流程（初始化 -> 查詢 -> 掛號 -> 滿號邊界 -> 取消）"""
    print("========== 開始執行主流程與完整流程測試 ==========\nInfo: 每次執行皆已重新初始化資料庫。\n")

    # 測試情境 1：查詢內科名額
    print("--- 測試情境 1：查詢內科名額 ---")
    print(process_clinic_request("內科還有名額嗎？", patient_name="張三"))
    print("-" * 50)

    # 測試情境 2：張三進行內科掛號
    print("--- 測試情境 2：張三進行內科掛號 ---")
    print(process_clinic_request("我要掛號內科", patient_name="張三"))
    print("-" * 50)

    # 測試情境 3：僅描述症狀（確保不提供醫療建議）
    print("--- 測試情境 3：僅描述症狀 ---")
    print(process_clinic_request("我最近發燒頭痛很嚴重", patient_name="張三"))
    print("-" * 50)

    # 測試情境 4：取消內科掛號
    print("--- 測試情境 4：取消內科掛號 ---")
    print(process_clinic_request("幫我取消內科", patient_name="張三"))
    print("-" * 50)

    print("\n完整流程測試完畢！")

run_integration_test()
````

````python
# ============================================================
# 第 8 格：Gradio 網頁
# 提醒：放在最後一格；GitHub 上看不到網頁，批改看的是第 5、7 格的結果
# ============================================================
# （把 AI 產生、你看過並理解的程式貼在這裡）
# (8) Gradio 網頁介面
import gradio as gr

def clinic_chatbot(user_input, patient_name):
    """Gradio 包裝函式"""
    if not user_input.strip():
        return "請輸入您的需求或問題。"
    if not patient_name.strip():
        patient_name = "訪客"
    return process_clinic_request(user_input, patient_name=patient_name)

# 建立 Gradio 介面
demo = gr.Interface(
    fn=clinic_chatbot,
    inputs=[
        gr.Textbox(label="您的需求 / 想說的話", placeholder="例如：我要掛號內科、內科還有名額嗎？、發燒了..."),
        gr.Textbox(label="患者姓名", value="王小明", placeholder="請輸入看診姓名")
    ],
    outputs=gr.Textbox(label="系統回應與 AI 解析"),
    title="🏥 智慧診所掛號系統 (Google Colab + SQLite + Groq)",
    description="本系統支援內科、外科、牙科之掛號、取消與名額查詢。AI 僅進行意圖與科別分類，不提供任何醫療建議。"
)

# 啟動網頁（在 Colab 中可以直接點擊產生的本地網址或 public link 測試）
demo.launch(inline=False)
````
